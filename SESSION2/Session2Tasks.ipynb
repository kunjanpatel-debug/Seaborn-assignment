```python
#task1

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df_swiggy = pd.read_csv('swiggy_payment_orders.csv')

plt.figure(figsize=(7, 4))
sns.countplot(data=df_swiggy, x='payment_method', palette='Blues_d')

plt.title('Number of Orders by Payment Method (Swiggy)')
plt.xlabel('Payment Method')
plt.ylabel('Order Count')
plt.tight_layout()
plt.show()
```

```python
#task2

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

df_zomato = pd.read_csv('zomato_delivery_times.csv')

plt.figure(figsize=(8, 4))
sns.barplot(data=df_zomato, x='food_category', y='delivery_time_min', estimator=np.mean, errorbar=None, palette='viridis')

plt.title('Average Delivery Time by Food Category')
plt.xlabel('Food Category')
plt.ylabel('Average Delivery Time (minutes)')
plt.tight_layout()
plt.show()
```

```python
#task3

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df_zomato = pd.read_csv('zomato_delivery_times.csv')

plt.figure(figsize=(8, 4))
sns.barplot(data=df_zomato, x='food_category', y='order_count', estimator=sum, errorbar=None, palette='mako')

plt.title('Total Orders by Food Category')
plt.xlabel('Food Category')
plt.ylabel('Total Orders')
plt.tight_layout()
plt.show()
```

```python
#task4

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df_bms = pd.read_csv('bookmyshow_movies.csv')

genre_counts = df_bms['genre'].value_counts().reset_index()
genre_counts.columns = ['genre', 'count']

plt.figure(figsize=(8, 4))
sns.barplot(data=genre_counts, x='genre', y='count', errorbar=None, palette='flare')

plt.title('BookMyShow Bookings by Genre (No Confidence Interval)')
plt.xlabel('Genre')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```
