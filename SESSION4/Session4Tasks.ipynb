```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#task1

df_fk = pd.read_csv('flipkart_prices.csv')

plt.figure(figsize=(6, 3.5))
sns.boxplot(data=df_fk, x='price', color='lightblue')
plt.title('Flipkart Product Price Distribution')
plt.xlabel('Price (₹)')
plt.tight_layout()
plt.show()
```

```python
#task2

df_zomato = pd.read_csv('zomato_orders_cat.csv')

plt.figure(figsize=(6, 3.5))
sns.violinplot(data=df_zomato, x='category', y='order_amount', hue='category', palette='pastel')
plt.title('Order Amount: Fast Food vs Desserts')
plt.xlabel('Category')
plt.ylabel('Order Amount (₹)')
plt.tight_layout()
plt.show()
```

```python
#task3

df_tickets = pd.read_csv('ipl_ticket_prices.csv')

plt.figure(figsize=(6, 3.5))
sns.boxplot(data=df_tickets, x='ticket_price', color='salmon')
plt.title('IPL Ticket Prices (Outlier Detection)')
plt.xlabel('Ticket Price (₹)')
plt.tight_layout()
plt.show()
```

```python
#task4

df_steps = pd.read_csv('fitness_steps_s4.csv')

plt.figure(figsize=(6, 3.5))
sns.violinplot(data=df_steps, y='daily_steps', color='lightgreen')
plt.title('Daily Step Count Distribution')
plt.ylabel('Daily Steps')
plt.tight_layout()
plt.show()
```
