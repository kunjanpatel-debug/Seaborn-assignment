```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#task1

df_insta = pd.read_csv('instagram_users.csv')

plt.figure(figsize=(6, 4))
sns.scatterplot(data=df_insta, x='posts', y='followers', color='teal', s=60)

plt.title('Instagram: Posts vs Followers')
plt.xlabel('Number of Posts')
plt.ylabel('Number of Followers')
plt.tight_layout()
plt.show()
```

```python
#task2

df_orders = pd.read_csv('zomato_daily_orders.csv')

plt.figure(figsize=(7, 4))
sns.lineplot(data=df_orders, x='day', y='orders', marker='o', color='crimson')

plt.title('Daily Zomato Food Orders (30 Days)')
plt.xlabel('Day of Month')
plt.ylabel('Order Count')
plt.tight_layout()
plt.show()
```

```python
#task3

df_movies = pd.read_csv('movie_reviews_genre.csv')

g = sns.relplot(data=df_movies, x='rating', y='reviews', hue='genre', s=90, height=4, aspect=1.3)

g.set_axis_labels('Movie Rating', 'Number of Reviews')
g.fig.suptitle('Movie Ratings vs Reviews by Genre', y=1.03)
plt.show()
```

```python
#task4

df_students = pd.read_csv('student_study_scores.csv')

plt.figure(figsize=(6, 4))
sns.regplot(data=df_students, x='youtube_hours', y='exam_score', color='darkblue', scatter_kws={'s': 50})

plt.title('YouTube Hours vs Exam Score')
plt.xlabel('Hours Spent on YouTube')
plt.ylabel('Exam Score (%)')
plt.tight_layout()
plt.show()
```
