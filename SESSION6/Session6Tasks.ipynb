```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#task1

df_fk = pd.read_csv('flipkart_rating_price.csv')

sns.jointplot(data=df_fk, x='rating', y='price', kind='scatter', color='royalblue')
plt.show()
```

```python
#task2

df_swiggy = pd.read_csv('swiggy_orders_joint.csv')

sns.jointplot(data=df_swiggy, x='delivery_time', y='order_value', kind='hex', color='darkorange')
plt.show()
```

```python
#task3

df_insta = pd.read_csv('instagram_posts_joint.csv')

sns.jointplot(data=df_insta, x='likes', y='comments', kind='kde', fill=True, color='purple')
plt.show()
```

```python
#task4

df_zomato = pd.read_csv('zomato_reviews_joint.csv')

sns.jointplot(data=df_zomato, x='rating', y='num_reviews', kind='scatter', color='teal')
plt.show()

sns.jointplot(data=df_zomato, x='rating', y='num_reviews', kind='hex', color='teal')
plt.show()

sns.jointplot(data=df_zomato, x='rating', y='num_reviews', kind='kde', fill=True, color='teal')
plt.show()
```
