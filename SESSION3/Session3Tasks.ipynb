```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#task1

df_swiggy = pd.read_csv('swiggy_delivery_times.csv')

plt.figure(figsize=(7, 4))
sns.histplot(data=df_swiggy, x='delivery_time_min', color='teal', edgecolor='black')

plt.title('Distribution of Swiggy Delivery Times')
plt.xlabel('Delivery Time (minutes)')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
#task2

df_steps = pd.read_csv('fitness_steps.csv')

plt.figure(figsize=(7, 4))
sns.kdeplot(data=df_steps, x='step_count', fill=True, color='purple')

plt.title('Daily Step Count Distribution (KDE)')
plt.xlabel('Steps per Day')
plt.ylabel('Density')
plt.tight_layout()
plt.show()
```

```python
#task3

df_zomato = pd.read_csv('zomato_monthly_spending.csv')

plt.figure(figsize=(7, 4))
sns.histplot(data=df_zomato, x='spending_amount', stat='density', color='skyblue', edgecolor='white')
sns.kdeplot(data=df_zomato, x='spending_amount', color='red', linewidth=2)

plt.title('Monthly Zomato Spending Distribution (Hist + KDE)')
plt.xlabel('Spending Amount (₹)')
plt.ylabel('Density')
plt.tight_layout()
plt.show()
```

```python
#task4

df_ipl = pd.read_csv('ipl_match_scores.csv')

for b in [5, 10, 20]:
    plt.figure(figsize=(6, 3.5))
    sns.histplot(data=df_ipl, x='total_runs', bins=b, color='coral', edgecolor='black')
    plt.title(f'IPL Scores Distribution (bins={b})')
    plt.xlabel('Total Runs')
    plt.ylabel('Frequency')
    plt.tight_layout()
    plt.show()
```
