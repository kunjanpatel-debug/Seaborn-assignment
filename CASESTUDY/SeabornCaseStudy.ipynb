```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
#task1

bank_df = pd.read_csv('bank_marketing.csv')

print(bank_df.head(10))

print(bank_df.info())
```

```python
#task2

plt.figure(figsize=(6, 4))
sns.countplot(data=bank_df, x='y', color='steelblue')

plt.title('Term Deposit Subscriptions (Count)')
plt.xlabel('Subscribed to Term Deposit (y)')
plt.ylabel('Number of Customers')
plt.tight_layout()
plt.show()
```

```python
#task3

plt.figure(figsize=(6, 4))
sns.boxplot(data=bank_df, x='y', y='age', color='lightcoral')

plt.title('Customer Age Distribution by Subscription Status')
plt.xlabel('Subscribed (y)')
plt.ylabel('Age')
plt.tight_layout()
plt.show()
```

```python
#task4

edu_perc = (
    bank_df.groupby('education')['y']
    .apply(lambda s: (s == 'yes').mean() * 100)
    .reset_index()
)
edu_perc.columns = ['education', 'subscription_rate']

plt.figure(figsize=(6, 4))
sns.barplot(
    data=edu_perc, x='education', y='subscription_rate', color='mediumseagreen'
)

plt.title('Subscription Rate by Education Level')
plt.xlabel('Education Level')
plt.ylabel('Subscription Rate (%)')
plt.tight_layout()
plt.show()
```

```python
#task5

#Two Recommended Visualizations:

#1.User Account Balance Distribution vs. Product Uptake (sns.boxplot): Fintech platforms segment audiences by liquid net worth. Users with higher account balances have idle funds that can be parked into a dedicated savings product.

#2.Existing Loan/Debt Status vs. Subscription (sns.countplot with hue):  Users with active housing or personal loans often allocate surplus capital toward EMIs rather than discretionary savings.

bank_df = pd.read_csv('bank_marketing.csv')

plt.figure(figsize=(7, 4.5))
sns.boxplot(data=bank_df, x='y', y='balance', color='plum')

plt.title('User Account Balance vs Product Subscription')
plt.xlabel('Subscribed to Savings Product (y)')
plt.ylabel('Account Balance (₹ / $)')
plt.tight_layout()
plt.show()
```
