```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#task1

df_insta = pd.read_csv('insta_posts_s9.csv')

plt.figure(figsize=(6, 4))
sns.regplot(data=df_insta, x='likes', y='comments', color='royalblue')

plt.title('Instagram: Likes vs Comments')
plt.xlabel('Likes')
plt.ylabel('Comments')
plt.tight_layout()
plt.show()
```

```python
#task2

df_zomato = pd.read_csv('zomato_orders_s9.csv')

g = sns.lmplot(data=df_zomato, x='order_price', y='delivery_time', height=4, aspect=1.3)

g.set_axis_labels('Order Price (₹)', 'Delivery Time (minutes)')
g.fig.suptitle('Zomato: Order Price vs Delivery Time', y=1.03)
plt.show()
```

```python
#task4

df_students = pd.read_csv('student_prep_s9.csv')

g = sns.lmplot(data=df_students, x='youtube_hours', y='exam_score', col='preparation_method', height=3.8, aspect=1.1)

g.set_axis_labels('YouTube Hours', 'Exam Score (%)')
g.fig.suptitle('YouTube Hours vs Exam Score by Preparation Method', y=1.05)
plt.show()
```

```python
#task5

df_fk = pd.read_csv('flipkart_reviews_s9.csv')

g = sns.lmplot(
    data=df_fk, 
    x='product_price', 
    y='num_reviews', 
    height=4, 
    aspect=1.3,
    scatter_kws={'color': 'crimson'},
    line_kws={'color': 'darkred'}
)

g.set_axis_labels('Product Price (₹)', 'Number of Reviews')
g.fig.suptitle('Flipkart: Product Price vs Number of Reviews', y=1.03)
plt.show()
```
