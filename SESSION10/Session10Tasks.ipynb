```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#task1

df_spotify = pd.read_csv('spotify_daily_songs.csv')

sns.set_theme(context='poster')

plt.figure(figsize=(9, 6))
sns.barplot(data=df_spotify, x='day', y='streams', color='teal')

plt.title('Daily Song Streams (Poster Context)')
plt.xlabel('Day')
plt.ylabel('Streams')
plt.tight_layout()
plt.show()
```

```python
#task2

sns.set_theme(context='notebook')
df_zomato = pd.read_csv('zomato_cuisine_orders.csv')

plt.figure(figsize=(8, 4.5))
sns.countplot(data=df_zomato, x='cuisine', hue='cuisine', palette='colorblind')
plt.legend([], [], frameon=False)

plt.title('Orders by Cuisine (Colorblind Palette)')
plt.xlabel('Cuisine Type')
plt.ylabel('Order Count')
plt.tight_layout()
plt.show()
```

```python
#task3

sns.set_theme(context='notebook', palette='muted')
df_steps = pd.read_csv('fitness_steps_s10.csv')

plt.figure(figsize=(8, 4.5))
sns.lineplot(data=df_steps, x='day', y='steps', marker='o', color='purple')

plt.title('Daily Step Counts (2 Weeks)')
plt.xlabel('Day')
plt.ylabel('Steps')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#task4

df_bms = pd.read_csv('bookmyshow_genre_orders.csv')

sns.set_theme(context='talk')
plt.figure(figsize=(8, 5))
sns.countplot(data=df_bms, x='genre', hue='genre', palette='deep')
plt.legend([], [], frameon=False)
plt.title('Before: talk context + deep palette')
plt.tight_layout()
plt.show()
```

```python
sns.set_theme(context='poster')
plt.figure(figsize=(10, 6.5))
sns.countplot(data=df_bms, x='genre', hue='genre', palette='colorblind')
plt.legend([], [], frameon=False)
plt.title('After: poster context + colorblind palette')
plt.tight_layout()
plt.show()
```

```python
#task5

df_flipkart = pd.read_csv('flipkart_category_sales.csv')

sns.set_theme(style='whitegrid', context='talk', font_scale=0.85)

plt.figure(figsize=(8, 4.5))
sns.barplot(data=df_flipkart, x='category', y='sales_amount', hue='category', palette='deep')
plt.legend([], [], frameon=False)

plt.title('Flipkart Sales Dashboard by Category')
plt.xlabel('Product Category')
plt.ylabel('Total Sales (₹)')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```
