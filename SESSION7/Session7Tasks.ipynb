```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#task1

df_delivery = pd.read_csv('food_delivery_features.csv')

corr_matrix = df_delivery.corr()

print(corr_matrix)
```

```python
#task2

plt.figure(figsize=(6, 4.5))
sns.heatmap(corr_matrix, annot=True, cmap='Blues')
plt.title('Food Delivery Feature Correlation (Annotated)')
plt.tight_layout()
plt.show()
```

```python
#task3

plt.figure(figsize=(6, 4.5))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Food Delivery Correlation (Diverging: coolwarm)')
plt.tight_layout()
plt.show()
```

```python
#task4

df_mobiles = pd.read_csv('flipkart_mobiles.csv')

corr_mobiles = df_mobiles.corr()

plt.figure(figsize=(6, 4.5))
sns.heatmap(corr_mobiles, annot=True, cmap='RdYlGn', vmin=-1, vmax=1)
plt.title('Flipkart Mobiles Feature Correlation (RdYlGn)')
plt.tight_layout()
plt.show()
```
