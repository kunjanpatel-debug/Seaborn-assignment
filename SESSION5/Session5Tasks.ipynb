```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#task1

tips = pd.read_csv('tips.csv')

sns.pairplot(tips)
plt.show()
```

```python
#task2

sns.pairplot(tips, hue='sex')
plt.show()
```

```python
#task3

sns.pairplot(tips, hue='smoker', diag_kind='kde')
plt.show()
```

```python
#task4

sns.pairplot(tips, vars=['total_bill', 'tip'], hue='time')
plt.show()
```
