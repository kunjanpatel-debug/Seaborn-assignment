```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#task1

df_food = pd.read_csv('food_orders_s11.csv')

plt.figure(figsize=(6, 4))
sns.barplot(data=df_food, x='category', y='orders', color='coral')

plt.title('Orders per Food Category')
plt.xlabel('Food Category')
plt.ylabel('Order Count')
plt.tight_layout()

plt.savefig('food_orders_300dpi.png', dpi=300)
plt.show()
```

```python
#task2

df_steps = pd.read_csv('weekly_steps_s11.csv')

plt.figure(figsize=(6, 4))
sns.lineplot(data=df_steps, x='day', y='steps', marker='o', color='teal')

plt.title('Weekly Step Counts')
plt.xlabel('Day')
plt.ylabel('Steps')
plt.tight_layout()

plt.savefig('weekly_steps.svg', format='svg')
plt.show()
```

```python
#task3

df_marks = pd.read_csv('student_marks_s11.csv')

sns.set_theme(style='darkgrid')

plt.figure(figsize=(6, 4))
sns.barplot(data=df_marks, x='subject', y='score', color='royalblue')

plt.title('Subject Exam Scores')
plt.xlabel('Subject')
plt.ylabel('Score (%)')
plt.tight_layout()

plt.savefig('custom_chart_1200dpi.png', dpi=1200)
plt.show()
```

```python
#task4

df_apps = pd.read_csv('app_usage_s11.csv')

def apply_presentation_theme():
    sns.set_theme(style='white', context='talk', font_scale=0.9)

apply_presentation_theme()
plt.figure(figsize=(6, 6))
colors = sns.color_palette('pastel')[0:5]

plt.pie(df_apps['screen_time_mins'], labels=df_apps['app'], autopct='%1.1f%%', colors=colors, startangle=140)
plt.title('Top 5 Most-Used Apps')
plt.tight_layout()

plt.savefig('app_usage_pie.png', dpi=300)
plt.show()
```

```python
#task5


corr_data = pd.read_csv('study_habits_corr_s11.csv', index_col=0)

plt.figure(figsize=(6, 4.5))
sns.heatmap(corr_data, annot=True, cmap='coolwarm', vmin=-1, vmax=1)

plt.title('Study Habits Correlation Heatmap')
plt.tight_layout()

plt.savefig('heatmap_1200dpi.svg', format='svg', dpi=1200)
plt.show()
```
