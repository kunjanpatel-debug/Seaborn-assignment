```python
#TASK2

import seaborn as sns
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

plt.figure(figsize=(7, 4))
sns.lineplot(x=days, y=temps, marker='o', color='crimson')

plt.title('Weekly Temperature Trend')
plt.xlabel('Day')
plt.ylabel('Temperature (°C)')

plt.show()
```

```python
#TASK3


days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

plt.figure(figsize=(6, 4))
sns.lineplot(x=days, y=temps, marker='o', color='crimson')
plt.title('Weekly Temperature Trend (Axes-level: lineplot)')
plt.xlabel('Day')
plt.ylabel('Temperature (°C)')
plt.show()
```

```python

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

g = sns.relplot(x=days, y=temps, kind='line', marker='o', color='teal', height=4, aspect=1.3)
g.set_axis_labels('Day', 'Temperature (°C)')
g.fig.suptitle('Weekly Temperature Trend (Figure-level: relplot)', y=1.03)
plt.show()
```

```python
#TASK4

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]
themes = ['darkgrid', 'whitegrid', 'ticks']

for theme in themes:
    sns.set_theme(style=theme)
    
    plt.figure(figsize=(6, 3.5))
    sns.lineplot(x=days, y=temps, marker='o')
    plt.title(f'Theme: {theme}')
    plt.xlabel('Day')
    plt.ylabel('Temperature (°C)')
    
    plt.savefig(f'theme_{theme}.png')
    plt.show()
```
